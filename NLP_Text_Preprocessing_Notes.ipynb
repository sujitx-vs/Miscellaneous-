## Basic Preprocessing
### Lowercasing

```
text='Hello WORLD!'
print(text.lower())
```

### Remove HTML Tags

```
import re
def remove_html_tags(text):
    return re.sub(r'<.*?>','',text)
print(remove_html_tags('<p>Hello</p>'))
```

### Remove URLs

```
def remove_url(text):
    return re.sub(r'https?://\S+|www\.\S+','',text)
print(remove_url('Visit https://google.com now'))
```

### Remove Punctuation

```
import string
exclude=string.punctuation
def remove_punc(text):
    return text.translate(str.maketrans('','',exclude))
print(remove_punc("Hello, World!"))
```

### Chat Word Conversion

```
chat_words={'IMO':'in my opinion','GN':'good night','POV':'point of view'}
def chat_conv(text):
    return ' '.join(chat_words.get(w.upper(),w) for w in text.split())
print(chat_conv('GN friends IMO'))
```

### Spelling Correction

```
from textblob import TextBlob
print(TextBlob('I havv a catt').correct())
```

### Remove Stopwords

```
from nltk.corpus import stopwords
sw=set(stopwords.words('english'))
def remove_stopwords(text):
    return ' '.join(w for w in text.split() if w.lower() not in sw)
print(remove_stopwords('this is a sample sentence'))
```

## Advanced Preprocessing
### Tokenization

```
from nltk.tokenize import word_tokenize,sent_tokenize
text='Hello world. NLP is fun!'
print(word_tokenize(text))
print(sent_tokenize(text))
```

### Stemming

```
from nltk.stem import PorterStemmer
ps=PorterStemmer()
print([ps.stem(w) for w in ['walk','walking','walked','walks']])
```

### Lemmatization

```
from nltk.stem import WordNetLemmatizer
lem=WordNetLemmatizer()
print([lem.lemmatize(w) for w in ['cars','wolves','running']])
```
